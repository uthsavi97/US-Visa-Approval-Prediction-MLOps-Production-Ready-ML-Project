<h2 style="color:#1F7A8C;">📘 About the Project</h2>

<p style="color:#2C3E50; font-size:15px; line-height:1.6;">
The <b>Immigration and Nationality Act (INA)</b> of the United States permits foreign workers 
to come to the country on either a <b>temporary</b> or <b>permanent</b> basis.  
The act also protects U.S. workers by ensuring fair hiring practices and maintaining 
requirements when employers recruit foreign talent.
</p>

<p style="color:#2C3E50; font-size:15px; line-height:1.6;">
These immigration programs are administered by the 
<b>Office of Foreign Labor Certification (OFLC)</b>.
</p>

<h2 style="color:#1F7A8C;">📌 1. Problem Statement</h2>

<p style="color:#2C3E50; font-size:15px; line-height:1.6;">
The <b>OFLC</b> processes job certification applications from employers 
seeking to bring foreign workers to the United States.
</p>

<p style="color:#2C3E50; font-size:15px; line-height:1.6;">
In recent years, the number of applications has increased significantly.  
To handle this large volume efficiently, the OFLC requires 
<b>Machine Learning models</b> to help shortlist visa applicants 
based on historical data.
</p>

<p style="color:#2C3E50; font-size:15px; line-height:1.6;">
In this project, we build a <b style="color:#27AE60;">Classification Model</b> that:
</p>

<ul style="color:#34495E; font-size:15px;">
  <li>✅ Predicts whether a visa will be <b>Approved</b> or <b>Denied</b></li>
  <li>📌 Recommends suitable applicant profiles</li>
  <li>📊 Identifies key factors influencing visa decisions</li>
</ul>
